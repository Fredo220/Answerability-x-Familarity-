Inspect the frozen test result, error taxonomy, prefix surfaces, and required ablations from saved artifacts.

```python
from pathlib import Path
import json
import pandas as pd
from trajectory_extractor import RunStore

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RUN_ID = 'concept-main'
store = RunStore(ROOT / 'runs')
metrics = json.loads((ROOT / 'runs' / RUN_ID / 'metrics' / 'detection.json').read_text())
pd.DataFrame({name: values['test'] for name, values in metrics['methods'].items()}).T
```

```python
runs = [store.read(RUN_ID, example_id) for example_id in store.example_ids(RUN_ID)]
pd.Series([run.provenance['concept_outcome'] for run in runs]).value_counts().rename('count')
```

```python
ablation_path = ROOT / 'runs' / RUN_ID / 'metrics' / 'ablations.json'
ablations = json.loads(ablation_path.read_text()) if ablation_path.exists() else {'status': 'run feature-dynamics ablate-concept'}
ablations
```

The primary label remains normalized exact-match error as preregistered. `concept_outcome` separately identifies true distractor-binding errors, formatting, refusal, truncation, and other failures.
