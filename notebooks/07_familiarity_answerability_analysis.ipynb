# Familiarity vs. Answerability: Sealed Analysis

This local notebook builds a report from closed live-store endpoints, then separately verifies an immutable release bundle. It never trains a probe, changes a threshold, or opens a protected endpoint.


```python
import json
import os
import subprocess
from pathlib import Path

from trajectory_extractor.fa_report import verify_release_bundle

ROOT = Path(os.environ.get("FA_RUN_ROOT", ".")).resolve()
BEHAVIOR_TEST_MANIFEST = os.environ["FA_BEHAVIOR_TEST_MANIFEST"]
PROBE_TEST_MANIFEST = os.environ["FA_PROBE_TEST_MANIFEST"]
SELECTION_MANIFEST = os.environ["FA_SELECTION_MANIFEST"]
```

```python
command = [
    "python", "-m", "trajectory_extractor.cli", "fa-build-report",
    "--config", "configs/familiarity_answerability_gemma2_2b.json",
    "--root", str(ROOT),
    "--behavior-test-manifest", BEHAVIOR_TEST_MANIFEST,
    "--probe-test-manifest", PROBE_TEST_MANIFEST,
    "--selection-manifest", SELECTION_MANIFEST,
    "--output", "reports/familiarity_answerability.md",
]
subprocess.run(command, check=True)
assert (ROOT / "reports/familiarity_answerability.md").is_file()
```

```python
RELEASE = Path(os.environ["FA_RELEASE_ROOT"]).resolve()
assert (RELEASE / "MANIFEST.json").is_file()
assert verify_release_bundle(RELEASE), "Release checksum verification failed."
manifest = json.loads((RELEASE / "MANIFEST.json").read_text(encoding="utf-8"))
manifest
```
